## 1. ¿Qué es Jupyter Notebook?

Jupyter Notebook es un entorno de computación interactiva que permite crear y compartir documentos con código, ecuaciones, visualizaciones y texto.

### Casos de uso principales:
- **Interactividad:** Permite la ejecución de código por celdas y la visualización inmediata de los resultados.
- **Visualización y Documentación:** Facilita la combinación de código con explicaciones, gráficas y otros elementos multimedia.
- **Ventajas:** Ideal para la exploración de datos, el prototipado rápido y la investigación reproducible.
- **Limitaciones:** No está diseñado para el desarrollo de software a gran escala o aplicaciones complejas.

````python
# Importamos librerías (cajas de herramientas)
import matplotlib.pyplot as plt

# Definimos datos (Meses y Ventas)
meses = ['Enero', 'Febrero', 'Marzo', 'Abril']
ventas = [10, 25, 15, 30]

# Creamos una gráfica simple
plt.figure(figsize=(8, 4))
plt.plot(meses, ventas, marker='o', linestyle='-', color='b')
plt.title('Crecimiento de Ventas')
plt.xlabel('Meses')
plt.ylabel('Cantidad')
plt.grid(True)
plt.show()

# Un cálculo rápido
promedio = sum(ventas) / len(ventas)
print(f"El promedio de ventas fue: {promedio}")
````

## 2. ¿Qué es GitHub y para qué se utiliza?

GitHub es una plataforma de desarrollo colaborativo basada en Git para alojar, revisar y gestionar proyectos de software.

### Conceptos clave:
- **Repositorio (Repository):** Contenedor del proyecto que almacena todos sus archivos y el historial de revisiones.
- **Commit:** Un punto de guardado que registra cambios específicos en los archivos del repositorio.
- **Rama (Branch):** Una línea de desarrollo independiente para trabajar en nuevas funcionalidades o experimentos sin afectar la rama principal.
- **Push:** Sincroniza los commits locales con el repositorio remoto en GitHub.
- **Pull Request (PR):** Propuesta para fusionar los cambios de una rama a otra, facilitando la revisión de código entre colaboradores.

### Comandos básicos de Git:

```bash
git clone <url-del-repositorio>
git pull
git branch
git checkout -b <nombre-rama>
git add .
git commit -m "Mensaje descriptivo"
git push origin <nombre-rama>


```

### Importancia de GitHub para equipos de ciencia de datos
GitHub permite a los equipos de ciencia de datos:
- Colaborar de manera eficiente en proyectos compartidos.
- Controlar versiones y mantener un historial claro de cambios.
- Revisar y validar código mediante pull requests.
- Integrar flujos de trabajo automatizados (CI/CD) para pruebas y despliegue.
- Facilitar la reproducibilidad y la transparencia en los análisis de datos.

## 3. Flujo de Trabajo Básico con Git y GitHub

1.  **Pull:** Sincronizar el repositorio local con los cambios más recientes del remoto.
    ```bash
    git pull
    ```
2.  **Desarrollo (Jupyter/IDE):** Implementar cambios, analizar datos o desarrollar nuevas funcionalidades.
3.  **Commit:** Guardar los cambios realizados en el repositorio local con un mensaje descriptivo.
    ```bash
    git add <archivo>
    git commit -m "Mensaje descriptivo"
    ```
4.  **Push:** Subir los commits al repositorio remoto.
    ```bash
    git push origin <nombre-rama>
    ```
5.  **Pull Request (Opcional):** Iniciar una solicitud de revisión para integrar los cambios en la rama principal desde la interfaz web de GitHub.
